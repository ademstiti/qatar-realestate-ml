```python
import pandas as pd

df = pd.read_csv("../data/raw_listings.csv")

print(df.shape)
print(df.dtypes)
print(df.isnull().sum())
df.head()
```

```python
#Clean price
df["price_qar"] = (
    df["price"]
    .str.replace(",", "")
    .str.replace(" QAR/month", "")
    .astype(int)
)

print(df["price_qar"].describe())
```

```python
#Clean area
df["area_sqm"] = (
    df["area"]
    .str.replace(" sqm", "")
    .astype(float)
)

print(df["area_sqm"].describe())
```

```python
#Clean bedrooms & bathrooms
df["bedrooms"]  = pd.to_numeric(df["bedrooms"],  errors="coerce")
df["bathrooms"] = pd.to_numeric(df["bathrooms"], errors="coerce")
```

```python
#Extract district and city
df["district"] = df["location"].str.split(",").str[-2].str.strip()
df["city"]     = df["location"].str.split(",").str[-1].str.strip()

print(df["district"].value_counts().head(10))
```

```python
#Drop duplicates, nulls, and messy columns
df = df.drop_duplicates()
df = df.dropna(subset=["price_qar", "area_sqm", "bedrooms"])
df = df.drop(columns=["price", "area"])

print(f"Clean dataset: {df.shape[0]} listings")
df.to_csv("../data/clean_listings.csv", index=False)
```
